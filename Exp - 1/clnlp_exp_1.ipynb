# Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('/content/drive/MyDrive/CLNLP - Lab/Exp - 1/employee_information_100.csv');

data.head()

avg_salary = data.groupby('Department')['Salary'].mean()

plt.figure(figsize=(6, 4))
avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)

plt.show()
```

# Experiment 1.2: Display the number of employees working in each department.

```python
dept_count = data["Department"].value_counts()

print(dept_count)

plt.figure(figsize=(6, 4))
dept_count.plot(kind="bar")

plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)

plt.show()
```

# Experiment 1.3: Display the percentage of male and female employees using a pie chart


```python
gender_count = data["Gender"].value_counts()

print(gender_count)

plt.figure(figsize=(5, 5))
plt.pie(gender_count, labels=gender_count.index, autopct="%1.1f%%", startangle=90)

plt.title("Percentage of Male and Female Employees")

plt.show()
```

# Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(6, 4))

plt.hist(data["Salary"], bins=5)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

# Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(6, 4))

plt.scatter(data["Experience_Years"], data["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

plt.show()
```

# Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top10 = data.sort_values(by="Salary", ascending=False).head(10)

top10
```

# Experiment 1.7: Find the highest salary in every department.

```python
highest_salary = data.groupby("Department")["Salary"].max()

highest_salary
```

# Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
avg_salary = data["Salary"].mean()

employees = data[data["Salary"] > avg_salary]

employees
```

# Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_experience = data.groupby("Department")["Experience_Years"].mean()

avg_experience
```
